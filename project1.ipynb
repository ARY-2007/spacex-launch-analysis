## Project Objective
The objective of this project is to perform a comprehensive data analysis of SpaceX orbital launch operations using the spacex_launches.csv dataset.

The primary goal is not only to visualize the data, but to extract meaningful insights from historical launch records and demonstrate how data analysis can be applied to the evaluation of aerospace mission operations.

## Data Preparation
In this section, the raw launch data is prepared and summarized to provide the information required for the subsequent analysis and visualizations.

### Extracting Launch Dates
The launch dates are extracted from the **"date_utc"** column to prepare the data for temporal analysis.

```python
date_list=list(df["date_utc"])
list_dates=[]
for i in date_list:
    date_utc_split=i.split("T")
    list_dates.append(date_utc_split[0])
date_df=pd.Series(list_dates)
time=pd.to_datetime(date_df)
a=list(time.dt.year)[0]
time_year_list=[a]
for j in list(time.dt.year):
    if j==a:
        pass
    else:
        time_year_list.append(j)
        a=j
```

### Calculating Annual Launch Counts
The total number of launches is calculated for each year to analyze changes in SpaceX launch activity over time.

```python
list_Number_of_launches_per_year=[]
for z in time_year_list:
    count=0
    for h in list(time.dt.year):
        if h==z:
            count+=1
        else:
            pass
    list_Number_of_launches_per_year.append(count)
```

### Calculating Overall Mission Success and Failure Rates
The overall percentages of successful and unsuccessful launches are calculated to evaluate the general reliability of SpaceX orbital missions.

```python
success_column=df["success"]
count_success=0
count_failure=0
for q in success_column:
    if q==1:
        count_success+=1
    else:
        count_failure+=1
total=len(success_column)
Success_rate=(count_success/total)*100
failure_rate=(count_failure/total)*100
```

### Calculating Annual Successful Launches
The number of successful launches is calculated for each year to examine how mission success has changed over time.

```python
list_success_time=[]
for l in time_year_list:
    new_df3=df[(df["success"]==1)&(time.dt.year==l)]
    list_success_time.append(len(new_df3))
```

### Identifying Rockets and Their Launch Distribution
The names of the rockets are identified, and the percentage of total launches performed by each rocket is calculated.

```python
rocket_names=df.groupby("rocket_name")
rocket_names_list=rocket_names.size().index.tolist()
counts_of_rockets =rocket_names.size().tolist()
Launch_Percentage=[(u1/len(df))*100 for u1 in counts_of_rockets]
```

### Calculating the Success Rate of Each Rocket
The success rate of each rocket is calculated to compare the reliability of different launch vehicles.

```python
rocket_success_rate_list=[]
for t in rocket_names_list:
    new_df=df[(df["rocket_name"]==t)&(df["success"]==1)]
    rocket_success_rate_list.append(len(new_df))

rocket_success_percentage_list=[(success_count / total_count) * 100 for success_count, total_count in zip(rocket_success_rate_list, counts_of_rockets)]
```

### Calculating Annual Booster Reuse
The number of reused rocket cores is determined for each year to analyze the development and use of booster reusability over time.

```python
our_time=pd.DataFrame(time.dt.year)
cores=pd.DataFrame(df["cores_reused"])
new_df_for_cores_time=pd.concat([our_time,cores],axis=1)
# Grouping by time
g1= new_df_for_cores_time.groupby(0)
result = g1["cores_reused"].sum()
reused_cores_per_year=result.tolist()
```

### Identifying Landing Types and Their Frequencies
The different booster landing types are identified, and the number of landings associated with each type is calculated.

```python
df_landing=df.groupby("landing_type")
landing_type_names=df_landing.size().index.tolist()
count_landing=df_landing.size().tolist()
```

### Calculating Successful Landings
The number of successful booster landings is calculated to evaluate overall landing performance.

```python
count_successful_landing_list=[]
for y2 in landing_type_names:
    our_new_df=df[(df["landing_type"]==y2)&(df["landing_success"]==1)]
    count_successful_landing_list.append(len(our_new_df))
```

### Analyzing Launchpad Activity
The launchpads used by SpaceX are identified, and the number of launches performed from each launchpad is calculated.

```python
df_launchpad_name=df.groupby("launchpad_name")
launchpad_names_list=df_launchpad_name.size().index.tolist()
count_launchpad=df_launchpad_name.size().tolist()
```

### Identifying Crewed and Uncrewed Missions
Launches are classified as crewed or uncrewed based on the number of crew members assigned to each mission.

```python
Manned_mission=len(df[df["crew_count"]!=0])
Unmanned_mission=len(df[df["crew_count"]==0])
```

### Calculating Annual Payload Activity
The number of payloads carried by SpaceX launches is calculated for each year to analyze changes in payload operations over time.

```python
Payload=df["payloads_count"]
payloads_time_df=pd.concat([time.dt.year,Payload],axis=1)
payload_group=payloads_time_df.groupby(0)
res=payload_group["payloads_count"].sum()
payloas_per_year=res.tolist()
```
